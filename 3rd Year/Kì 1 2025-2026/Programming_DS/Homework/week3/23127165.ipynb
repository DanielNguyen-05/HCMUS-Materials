### NumPy Exercises: Student Performance Dataset

#### 📋 Dataset Information

**Dataset**: Students Performance in Exams  
**Source**: https://www.kaggle.com/datasets/spscientist/students-performance-in-exams  
**Size**: 1,000 students  

**Columns**:
- `gender`: Student's gender (male/female)
- `race/ethnicity`: Student's ethnic group (group A, B, C, D, E)
- `parental level of education`: Parent's education level
- `lunch`: Type of lunch (standard/free or reduced)
- `test preparation course`: Completed test prep course (completed/none)
- `math score`: Score in math (0-100)
- `reading score`: Score in reading (0-100)
- `writing score`: Score in writing (0-100)

```python
## NO NEED CODE HERE
import numpy as np

# Load the dataset
data = np.genfromtxt('Data/StudentsPerformance.csv', delimiter=',', 
                     dtype=None, encoding='utf-8', names=True)

# For easier manipulation, separate numeric scores into a 2D array
math_scores = data['math_score']
reading_scores = data['reading_score']
writing_scores = data['writing_score']

# Create a 2D array with all scores
scores = np.column_stack([math_scores, reading_scores, writing_scores])

print("Dataset loaded successfully!")
print(f"Number of students: {len(data)}")
print(f"Scores array shape: {scores.shape}")
```

### Task 3: Filter Students by Performance
**Instructions**:
1. Find all students who scored above 80 in math
2. Find all students who scored below 50 in reading
3. Find all students who scored exactly 75 in writing
4. Count how many students scored above 90 in math
5. Find students who scored between 60 and 70 (inclusive) in reading

```python
# TODO: Your code here

# Convert scores to numeric by stripping quotes and converting to int
scores_numeric = np.array([[int(str(x).replace('"', '')) for x in row] for row in scores])
scores = scores_numeric

# 1. Students with math > 80
math_above_80_idx = np.where(scores[:, 0] > 80)[0] + 1
print("=> Students with math > 80:")
print(math_above_80_idx, "\n")

# 2. Students with reading < 50
reading_below_50_idx = np.where(scores[:, 1] < 50)[0] + 1
print("=> Students with reading < 50:")
print(reading_below_50_idx, "\n")

# 3. Students with writing exactly 75
writing_75_idx = np.where(scores[:, 2] == 75)[0] + 1
print("=> Students with writing = 75:")
print(writing_75_idx, "\n")

# 4. Count students with math > 90
count_math_above_90 = np.sum(scores[:, 0] > 90)
print(f"=> Number of students with math > 90: {count_math_above_90}\n")

# 5. Students with reading between 60-70 inclusive
reading_60_70_idx = np.where((scores[:, 1] >= 60) & (scores[:, 1] <= 70))[0] + 1
print("=> Students with reading between 60 and 70:")
print(reading_60_70_idx, "\n")
```

**Instructions**:

```python
# TODO: Your code here
# 1. Indices of students with math > reading
math_gt_reading_idx = np.where(scores[:, 0] > scores[:, 1])[0] + 1
print("=> Students with math > reading:")
print(math_gt_reading_idx, "\n")

# 2. Indices of students with reading == writing
reading_eq_writing_idx = np.where(scores[:, 1] == scores[:, 2])[0] + 1
print("=> Students with reading == writing:")
print(reading_eq_writing_idx, "\n")

# 3. Count students with writing > math
count_writing_gt_math = np.sum(scores[:, 2] > scores[:, 0]) 
print(f"=> Number of students with writing > math: {count_writing_gt_math}\n")

# 4. Student(s) with the largest difference between math and writing scores
diff_math_writing = np.abs(scores[:, 0] - scores[:, 2])
max_diff = np.max(diff_math_writing)
max_diff_idx = np.where(diff_math_writing == max_diff)[0] + 1
print(f"=> Student(s) with largest math-writing difference ({max_diff} points):")
print(max_diff_idx, "\n")

# 5. Average difference between reading and writing scores
avg_diff_reading_writing = np.mean(np.abs(scores[:, 1] - scores[:, 2]))
print(f"=> Average difference between reading and writing: {avg_diff_reading_writing:.2f}\n")
```

### Task 5: Score Distribution Analysis 
Analyze how scores are distributed across different ranges without using histograms or binning functions

**Instructions**:

1. **Create score bins**: Define 10 ranges for scores:
   - Bin 0: 0-9 points
   - Bin 1: 10-19 points
   - Bin 2: 20-29 points
   - ... and so on ...
   - Bin 9: 90-100 points

2. **Count students in each bin for math scores**: Use a loop to go through all math scores and count how many fall into each bin.

3. **Display the distribution**: Print the count for each bin in a readable format.

4. **Find the most common range**: Determine which bin has the most students (the mode range).

5. **Calculate cumulative distribution**: For each bin, calculate how many students scored at or below that range.

6. **Repeat for reading and writing**: Apply the same analysis to the other two subjects.

7. **Compare distributions**: Identify which subject has the most students in the highest bin (90-100).

```python
#TODO: your code here

subjects = ['Math', 'Reading', 'Writing']

for i, subj in enumerate(subjects):
    subj_scores = scores[:, i]
    
    # 1. Create an array to store counts for 10 bins
    bins = np.zeros(10, dtype=int)
    
    # 2. Count students in each bin
    for score in subj_scores:
        bin_idx = min(score // 10, 9)  # Ensure 100 goes into last bin
        bins[bin_idx] += 1
    
    # 3. Display the distribution
    print(f"\n{subj} Score Distribution:")
    for b in range(10):
        print(f"  {b*10}-{b*10+9}: {bins[b]} students")
    
    # 4. Find the most common range
    most_common_bin = np.argmax(bins)
    print(f"=> Most common range for {subj}: {most_common_bin*10}-{most_common_bin*10+9}")
    
    # 5. Calculate cumulative distribution
    cumulative = np.cumsum(bins)
    print(f"=> Cumulative distribution for {subj}: {cumulative}")

# 7. Compare highest bin (90-100)
highest_bin_counts = [np.sum(scores[:, i] >= 90) for i in range(3)]
subject_most_90_100 = subjects[np.argmax(highest_bin_counts)]
print(f"\n=> Subject with most students scoring 90-100: {subject_most_90_100}")
```

### Task 6: Student Consistency Analysis
Identify students who have very consistent scores across all three subjects without using variance or standard deviation functions.

**Instructions**:

1. **Calculate score range for each student**: For each student, find the difference between their highest and lowest score among the three subjects.

2. **Define consistency threshold**: Students are "consistent" if their range is 10 points or less (all three scores within 10 points of each other).

3. **Count consistent students**: Count how many students meet the consistency criteria.

4. **Find the most consistent student**: Identify the student with the smallest range (most consistent).

5. **Find the least consistent student**: Identify the student with the largest range (most variable).

6. **Create consistency categories**:
   - Very Consistent: range ≤ 5
   - Consistent: range 6-10
   - Moderate: range 11-20
   - Variable: range 21-30
   - Highly Variable: range > 30

7. **Count students in each category**: Display the distribution.

8. **Analyze patterns**: Among consistent students, which subject combinations tend to be similar?

```python
# TODO: Your code here

# 1. Calculate score range for each student
score_range = np.max(scores, axis=1) - np.min(scores, axis=1)

# 2 & 3. Define consistent students (range <= 10) and count
consistent_students_idx = np.where(score_range <= 10)[0]
print("=> Number of consistent students (range ≤ 10):", len(consistent_students_idx))

# 4. Find the most consistent student(s) (smallest range)
min_range = np.min(score_range)
most_consistent_idx = np.where(score_range == min_range)[0]
print("=> Most consistent student(s) indices:", most_consistent_idx)

# 5. Find the least consistent student(s) (largest range)
max_range = np.max(score_range)
least_consistent_idx = np.where(score_range == max_range)[0]
print("=> Least consistent student(s) indices:", least_consistent_idx)

# 6. Create consistency categories
categories = {
    "V=> ery Consistent (≤5)": np.where(score_range <= 5)[0],
    "=> Consistent (6-10)": np.where((score_range >= 6) & (score_range <= 10))[0],
    "=> Moderate (11-20)": np.where((score_range >= 11) & (score_range <= 20))[0],
    "=> Variable (21-30)": np.where((score_range >= 21) & (score_range <= 30))[0],
    "=> Highly Variable (>30)": np.where(score_range > 30)[0]
}

# 7. Count students in each category
for cat, idxs in categories.items():
    print(f"{cat}: {len(idxs)} students")

# 8. Analyze patterns among consistent students
consistent_scores = scores[consistent_students_idx]

# Differences between subject pairs among consistent students
math_reading_diff = np.abs(consistent_scores[:, 0] - consistent_scores[:, 1])
math_writing_diff = np.abs(consistent_scores[:, 0] - consistent_scores[:, 2])
reading_writing_diff = np.abs(consistent_scores[:, 1] - consistent_scores[:, 2])

print("\n=> Average differences among consistent students:")
print("\t+ Math-Reading:", np.mean(math_reading_diff))
print("\t+ Math-Writing:", np.mean(math_writing_diff))
print("\t+ Reading-Writing:", np.mean(reading_writing_diff))
```
