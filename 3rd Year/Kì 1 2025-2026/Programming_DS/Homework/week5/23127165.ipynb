# Pandas Hands-on Exercise - California Baby Names

### Dataset Introduction
This dataset (attached) contains information about **baby names** registered in California (USA) from 1910 to 2022.

#### Data Structure:
- **State**: State (CA = California)
- **Sex**: Gender (F = Female, M = Male)
- **Year**: Year of birth
- **Name**: Name
- **Count**: The number of babies with that name in that year

#### Example:
```
CA,F,2020,Olivia,2353  → In 2020, there were 2,353 baby girls named Olivia
CA,M,2020,Noah,2458    → In 2020, there were 2,458 baby boys named Noah
```

### Task 2: Trend Analysis: analyzing how the Name "Emma" change over time
**Instruction**
- Get all records for the name "Emma"
- In which year did Emma first appear?
- In which year was Emma most popular
- How many baby boys were named Emma?
- Calculate the total number of babies named Emma from 1910 to 2022.

````python
#TODO: your code here

name_to_analyze = 'Emma'

# 1. Get all records for the name "Emma"
emma_df = babynames_df[babynames_df['Name'] == name_to_analyze]
print(f"1. Filtered records for '{name_to_analyze}' with total records: {len(emma_df)}")
print(f"=> Records for the name '{name_to_analyze}':")
print(emma_df)

if not emma_df.empty:
    
    # 2. In which year did Emma first appear?
    first_appearance_year = emma_df['Year'].min()
    print(f"\n2. Year '{name_to_analyze}' first appeared: {first_appearance_year}")

    # 3. In which year was Emma most popular
    most_popular_record = emma_df.loc[emma_df['Count'].idxmax()]
    most_popular_year = most_popular_record['Year']
    most_popular_count = most_popular_record['Count']
    print(f"3. Year '{name_to_analyze}' was most popular: {most_popular_year} (Count: {most_popular_count:,})")

    # 4. How many baby boys were named Emma?
    boys_named_emma_count = emma_df[emma_df['Sex'] == 'M']['Count'].sum()
    print(f"4. Total baby boys named '{name_to_analyze}': {boys_named_emma_count:,}")

    # 5. Calculate the total number of babies named Emma from 1910 to 2022.
    total_emma_count = emma_df['Count'].sum()
    print(f"5. Total babies named '{name_to_analyze}' (1910-2022): {total_emma_count:,}")
else:
    print(f"Name '{name_to_analyze}' not found in the dataset.")
````

### Task 3: Analysis by Decade (1910s, 1920s, ..., 2020s)
**Instruction**
- Calculate the total number of babies born in each decade (create Decade column).
- Which decade had the highest number of babies?
- In the 1990s decade, how many unique names were there?
- "Very Popular" name mean Count >= 500. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
- Calculate the average Count for each decade.

````python
#TODO: your code here

# Create Decade column
babynames_df['Decade'] = (babynames_df['Year'] // 10) * 10

# 1. Calculate the total number of babies born in each decade.
births_by_decade = babynames_df.groupby('Decade')['Count'].sum()
print("1. Total number of babies born in each decade:")
print(births_by_decade)

print("-" * 80)

# 2. Which decade had the highest number of babies?
highest_birth_decade = births_by_decade.idxmax()
highest_birth_count = births_by_decade.max()
print(f"2. Decade with the highest number of babies: {highest_birth_decade}s (Total: {highest_birth_count:,})")

print("-" * 80)

# 3. In the 1990s decade, how many unique names were there?
decade_1990s = 1990
unique_names_1990s = babynames_df[babynames_df['Decade'] == decade_1990s]['Name'].nunique()
print(f"3. Unique names in the {decade_1990s}s decade: {unique_names_1990s}")

print("-" * 80)

# 4. Compare the number of "Very Popular" names between the 1980s and 2010s decades.
# "Very Popular" name means Count >= 500
very_popular_df = babynames_df[babynames_df['Count'] >= 500]

# Calculate unique names in each decade satisfying the count condition
vp_names_1980s = very_popular_df[very_popular_df['Decade'] == 1980]['Name'].nunique()
vp_names_2010s = very_popular_df[very_popular_df['Decade'] == 2010]['Name'].nunique()

print("4. Number of 'Very Popular' names (Count >= 500):")
print(f"\t- 1980s: {vp_names_1980s} unique names")
print(f"\t- 2010s: {vp_names_2010s} unique names")

print("-" * 80)

# 5. Calculate the average Count for each decade.
avg_count_by_decade = babynames_df.groupby('Decade')['Count'].mean()
print("5. Average Count for each decade:")
print(avg_count_by_decade)
````

### Task 4: Analyze `Timeless` Names (Unchanging Over Time)
**Definition**: A name is called `timeless` if it:
- Appeared in **ALL** decades from the 1910s to the 2020s
- Always had a Count $\ge$ 100 in every decade

**Instruction**
- Find all "timeless" names.
- How many such names are there?
- Display the list of these names.
- Among these names, which one has the highest total Count?

````python
#TODO: your code here

# The required decades are from the min to the max decade in the dataset (e.g., 1910 to 2020)
min_decade = babynames_df['Decade'].min()
max_decade = babynames_df['Decade'].max()
target_decades = np.arange(min_decade, max_decade + 10, 10)
num_target_decades = len(target_decades)

# 1. Summarize Count by Name and Decade (Sum of Count across Sexes/Years within a decade)
name_decade_summary = babynames_df.groupby(['Name', 'Decade'])['Count'].sum().reset_index()

# A. Find names that always had a Count >= 100 in EVERY decade
# (Filter names that had Count < 100 in AT LEAST one decade)
infringing_names_df = name_decade_summary[name_decade_summary['Count'] < 100]
infringing_names = infringing_names_df['Name'].unique()
    
# B. Find names that appeared in ALL target decades
name_decade_counts = name_decade_summary.groupby('Name')['Decade'].nunique()
fully_appearing_names = name_decade_counts[name_decade_counts == num_target_decades].index
    
# Timeless Names: Appeared fully AND did not infringe the Count >= 100 rule
timeless_names = fully_appearing_names[~fully_appearing_names.isin(infringing_names)]

# 1. & 2. Find all "timeless" names and count them
num_timeless_names = len(timeless_names)
print(f"1. & 2. Number of 'Timeless' Names: {num_timeless_names}")

# 3. Display the list of these names.
print(f"\n3. List of 'Timeless' Names:")
print(timeless_names.tolist())

# 4. Among these names, which one has the highest total Count?
timeless_df = babynames_df[babynames_df['Name'].isin(timeless_names)]
total_count_timeless = timeless_df.groupby('Name')['Count'].sum()

if not total_count_timeless.empty:
    highest_count_name = total_count_timeless.idxmax()
    highest_count = total_count_timeless.max()
    print(f"\n4. 'Timeless' Name with the highest total Count: {highest_count_name} (Total Count: {highest_count:,})")
else:
    print("\n4. No 'Timeless' names found to compare total Count.")
````

### Task 5: `Your Name` Analysis
**Perform an in-depth analysis of any given name**\
Create a function `analyze_name(name)` that performs the following:

1. Check if the name exists in the dataset.
2. If it exists, display:
   - The first year it appeared.
   - The most popular year (and its Count).
   - The total number of people with this name.
   - Which gender uses this name more frequently.
   - Which category this name belongs to (Rare/Uncommon/Common/Very Popular) in the last 5 years.
     
***Test with these name:***
```
['Emma', 'Noah', 'Michael', 'Sophia', 'James']
```

````python
#TODO: your code here

def get_popularity_category(count):
    """Classifies popularity based on Count."""
    if count < 10:
        return "Rare"
    elif 10 <= count < 100:
        return "Uncommon"
    elif 100 <= count < 500:
        return "Common"
    else: # Count >= 500
        return "Very Popular"

def analyze_name(name):
    print(f"\t--- ANALYSIS FOR NAME: {name} ---")
    name_df = babynames_df[babynames_df['Name'] == name]

    # 1. Check if the name exists
    if name_df.empty:
        print(f"1. Name '{name}' does not exist in the dataset.")
        return

    # 2. Display information
    
    # a. The first year it appeared.
    first_year = name_df['Year'].min()
    print(f"2a. First year appeared: {first_year}")

    # b. The most popular year (and its Count).
    most_popular_record = name_df.loc[name_df['Count'].idxmax()]
    most_popular_year = most_popular_record['Year']
    most_popular_count = most_popular_record['Count']
    print(f"2b. Most popular year: {most_popular_year} (Count: {most_popular_count})")

    # c. The total number of people with this name.
    total_count = name_df['Count'].sum()
    print(f"2c. Total number of people named '{name}': {total_count:,}")

    # d. Which gender uses this name more frequently.
    gender_counts = name_df.groupby('Sex')['Count'].sum()
    dominant_gender = gender_counts.idxmax()
    print(f"2d. Dominant gender: {dominant_gender}")

    # e. Which category this name belongs to in the last 5 years.
    last_5_years_start = babynames_df['Year'].max() - 4
    last_5_years_df = name_df[name_df['Year'] >= last_5_years_start]
    
    # Use the total Count in the last 5 years for classification
    last_5_years_count_sum = last_5_years_df['Count'].sum()
    category = get_popularity_category(last_5_years_count_sum)
    
    print(f"2e. Popularity in the last 5 years (Total Count since {last_5_years_start}): {category} ({last_5_years_count_sum:,})")
    print("-" * 80)
````

````python
test_names = ['Emma', 'Noah', 'Michael', 'Sophia', 'James']

for name in test_names:
    analyze_name(name)
````
