## **Linear Regression Error**

### **Problem 1**
**Question:** For $\sigma = 0.1$ and $d = 8$, find the smallest $N$ such that the expected in-sample error $\mathbb{E}[E_{in}]$ is greater than 0.008.

**Analysis:**
The expected in-sample error is given by the formula:
$$ \mathbb{E}_{\mathcal{D}}[E_{in}(\mathbf{w}_{lin})] = \sigma^2 \left(1 - \frac{d+1}{N}\right) $$

Given:
- $\sigma = 0.1 \implies \sigma^2 = 0.01$
- $d = 8$
- We need $\mathbb{E}[E_{in}] > 0.008$

Substitute the values:
$$ 0.01 \left(1 - \frac{8+1}{N}\right) > 0.008 $$
$$ 0.01 \left(1 - \frac{9}{N}\right) > 0.008 $$

Divide both sides by 0.01:
$$ 1 - \frac{9}{N} > 0.8 $$

Rearrange to solve for $N$:
$$ 0.2 > \frac{9}{N} $$
$$ N > \frac{9}{0.2} $$
$$ N > 45 $$

We look for the smallest option that satisfies $N > 45$.
- [a] 10 (False)
- [b] 25 (False)
- [c] 100 (True, smallest valid)
- [d] 500 (True)
- [e] 1000 (True)

**Answer: [c] 100**

## **Nonlinear Transforms**

### **Problem 2**
**Question:** Which constraints on $\tilde{w}$ correspond to the hyperbolic decision boundary depicted?

**Analysis:**
The transform is $\Phi(1, x_1, x_2) = (1, x_1^2, x_2^2)$.
The linear model in $\mathcal{Z}$-space is $h(\mathbf{x}) = \text{sign}(\tilde{w}_0 + \tilde{w}_1 x_1^2 + \tilde{w}_2 x_2^2)$.
The decision boundary is $\tilde{w}_0 + \tilde{w}_1 x_1^2 + \tilde{w}_2 x_2^2 = 0$.

Looking at the figure:
1.  **Origin:** The region containing $(0,0)$ is labeled **+1**.
    Plugging in $(0,0)$: $\text{sign}(\tilde{w}_0) = +1 \implies \tilde{w}_0 > 0$.
2.  **$x_1$-axis:** Moving far along the $x_1$ axis (setting $x_2=0$), we cross the boundary into the **-1** region.
    So for large $x_1$, $\tilde{w}_0 + \tilde{w}_1 x_1^2$ must be negative. Since $x_1^2 > 0$ and $\tilde{w}_0 > 0$, $\tilde{w}_1$ must be negative to pull the sum below zero.
    $\implies \tilde{w}_1 < 0$.
3.  **$x_2$-axis:** Moving far along the $x_2$ axis (setting $x_1=0$), we stay in the central **+1** region (between the two hyperbolic branches).
    So for large $x_2$, $\tilde{w}_0 + \tilde{w}_2 x_2^2$ must be positive.
    $\implies \tilde{w}_2 > 0$.

**Answer: [d] $\tilde{w}_1 < 0, \tilde{w}_2 > 0$**

### **Problem 3**
**Question:** What is the VC dimension of a linear model in the space transformed by a 4th-order polynomial?

**Analysis:**
We need to count the number of dimensions (features) in the transformed space $\tilde{d}$. The VC dimension of a linear model (Perceptron) in $\tilde{d}$ dimensions is roughly $\tilde{d}$ (specifically $\tilde{d}+1$ if counting bias as a dimension, or exactly the number of weights including bias).

Let's list the terms for a 4th-order polynomial in 2 variables ($x_1, x_2$):
- Order 0: $1$ (1 term)
- Order 1: $x_1, x_2$ (2 terms)
- Order 2: $x_1^2, x_1x_2, x_2^2$ (3 terms)
- Order 3: $x_1^3, x_1^2x_2, x_1x_2^2, x_2^3$ (4 terms)
- Order 4: $x_1^4, x_1^3x_2, x_1^2x_2^2, x_1x_2^3, x_2^4$ (5 terms)

Total dimensions = $1 + 2 + 3 + 4 + 5 = 15$.
The VC dimension $d_{VC} \le 15$. Since it's a linear model in this space, $d_{VC} = 15$.
The question asks for the smallest value *not smaller* than the VC dimension. That is 15.

**Answer: [c] 15**

## **Gradient Descent**

**Given:**
$E(u, v) = (ue^v - 2ve^{-u})^2$.
Initial point $(u, v) = (1, 1)$, $\eta = 0.1$.



### **Problem 4**
**Question:** Calculate $\frac{\partial E}{\partial u}$.

**Analysis:**
Let $A = ue^v - 2ve^{-u}$. Then $E = A^2$.
Using the chain rule: $\frac{\partial E}{\partial u} = 2A \cdot \frac{\partial A}{\partial u}$.

Calculate $\frac{\partial A}{\partial u}$:
$\frac{\partial}{\partial u} (ue^v - 2ve^{-u}) = e^v - 2v(-e^{-u}) = e^v + 2ve^{-u}$.

Substituting back:
$\frac{\partial E}{\partial u} = 2(ue^v - 2ve^{-u})(e^v + 2ve^{-u})$.

**Answer: [e] $2(e^v + 2ve^{-u})(ue^v - 2ve^{-u})$**

### **Problems 5, 6, 7 (Computational)**
We write a Python script to compute the iterations and values.

```python
import numpy as np

def get_error(u, v):
    return (u * np.exp(v) - 2 * v * np.exp(-u))**2

def get_gradient(u, v):
    term = u * np.exp(v) - 2 * v * np.exp(-u)
    dE_du = 2 * term * (np.exp(v) + 2 * v * np.exp(-u))
    dE_dv = 2 * term * (u * np.exp(v) - 2 * np.exp(-u))
    return dE_du, dE_dv

# --- Problem 5 & 6: Gradient Descent ---
u, v = 1.0, 1.0
eta = 0.1
iterations = 0

while True:
    error = get_error(u, v)
    if error < 1e-14:
        break
    du, dv = get_gradient(u, v)
    u -= eta * du
    v -= eta * dv
    iterations += 1

print(f"Problem 5: Iterations = {iterations}")
print(f"Problem 6: Final (u, v) = ({u:.3f}, {v:.3f})")

# --- Problem 7: Coordinate Descent ---
u, v = 1.0, 1.0
eta = 0.1
for _ in range(15):
    # Step 1: Update u
    du, _ = get_gradient(u, v) # Partial wrt u
    u -= eta * du
    
    # Step 2: Update v (using new u)
    _, dv = get_gradient(u, v) # Partial wrt v
    v -= eta * dv

final_error = get_error(u, v)
print(f"Problem 7: Final Error = {final_error:.1e}")
```

**Answers:**
- **Problem 5: [d] 10**
- **Problem 6: [e] (0.045, 0.024)**
- **Problem 7: [a] $10^{-1}$**

## **Logistic Regression**

### **Problems 8, 9 (Computational)**
We simulate Logistic Regression with Stochastic Gradient Descent (SGD).

**Simulation details:**
- $N=100$.
- Target: Line defined by 2 random points in $[-1, 1]^2$.
- Features: $[1, x_1, x_2]$.
- SGD Learning Rate: $0.01$.
- Stop: $\|\mathbf{w}^{(t-1)} - \mathbf{w}^{(t)}\| < 0.01$.
- Runs: 100.

```python
import numpy as np

def generate_data(N):
    return np.random.uniform(-1, 1, (N, 2))

def run_simulation():
    runs = 100
    total_epochs = 0
    total_e_out = 0
    
    for _ in range(runs):
        # 1. Target Function
        p = np.random.uniform(-1, 1, (2, 2))
        # Line passing through p1, p2
        # Normal vector w_f calculation
        w_f = np.array([p[1,0]*p[0,1] - p[0,0]*p[1,1], p[1,1]-p[0,1], p[0,0]-p[1,0]])
        
        # 2. Training Data
        X = np.c_[np.ones(100), generate_data(100)]
        y = np.sign(X @ w_f)
        
        # 3. Logistic Regression SGD
        w = np.zeros(3)
        eta = 0.01
        epoch = 0
        
        while True:
            w_old = w.copy()
            perms = np.random.permutation(100)
            
            for i in perms:
                xn = X[i]
                yn = y[i]
                # Gradient of Cross Entropy: -yn*xn / (1 + exp(yn*w^T*xn))
                grad = -yn * xn / (1 + np.exp(yn * np.dot(w, xn)))
                w -= eta * grad
            
            epoch += 1
            if np.linalg.norm(w - w_old) < 0.01:
                break
        
        total_epochs += epoch
        
        # 4. E_out estimation
        X_test = np.c_[np.ones(1000), generate_data(1000)]
        y_test = np.sign(X_test @ w_f)
        # Cross Entropy Error
        scores = X_test @ w
        # ln(1 + exp(-y * score))
        e_out = np.mean(np.log(1 + np.exp(-y_test * scores)))
        total_e_out += e_out

    print(f"Problem 8: Avg E_out = {total_e_out/runs:.3f}")
    print(f"Problem 9: Avg Epochs = {total_epochs/runs:.0f}")

run_simulation()
```

**Answers:**
- **Problem 8: [d] 0.100**
- **Problem 9: [a] 350**

## **PLA as SGD**

### **Problem 10**
**Question:** Which error function $e_n(\mathbf{w})$ makes SGD equivalent to the Perceptron Learning Algorithm (PLA)?

**Analysis:**
Let's look at the update rules.
**SGD Update:** $\mathbf{w}(t+1) = \mathbf{w}(t) - \eta \nabla e_n(\mathbf{w})$.
Assuming $\eta=1$.

**PLA Update:**
Given a point $(\mathbf{x}_n, y_n)$:
- Case 1: Correctly classified ($y_n \mathbf{w}^T \mathbf{x}_n > 0$).
  Update: $\mathbf{w} \leftarrow \mathbf{w}$ (No change).
  This implies $\nabla e_n(\mathbf{w}) = 0$.
- Case 2: Misclassified ($y_n \mathbf{w}^T \mathbf{x}_n \le 0$).
  Update: $\mathbf{w} \leftarrow \mathbf{w} + y_n \mathbf{x}_n$.
  Matching this with SGD: $\mathbf{w} + y_n \mathbf{x}_n = \mathbf{w} - \nabla e_n(\mathbf{w})$.
  $\implies \nabla e_n(\mathbf{w}) = -y_n \mathbf{x}_n$.

Now let's check the derivative of option [e]:
$$ e_n(\mathbf{w}) = -\min(0, y_n \mathbf{w}^T \mathbf{x}_n) $$
Let $s = y_n \mathbf{w}^T \mathbf{x}_n$. Then $e_n = -\min(0, s)$.
- If $s > 0$ (Correct): $\min(0, s) = 0 \implies e_n = 0$. Derivative is 0. (Matches Case 1).
- If $s < 0$ (Misclassified): $\min(0, s) = s \implies e_n = -s = -y_n \mathbf{w}^T \mathbf{x}_n$.
  Derivative with respect to $\mathbf{w}$ is $-y_n \mathbf{x}_n$. (Matches Case 2).

**Answer: [e] $e_n(\mathbf{w}) = -\min(0, y_n \mathbf{w}^T \mathbf{x}_n)$**
