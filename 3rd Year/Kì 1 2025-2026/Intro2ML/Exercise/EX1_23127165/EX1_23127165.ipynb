## **Question 3:**  

This is a conditional probability problem.  
Let $B_1$ be the event of selecting Bag 1 (two black balls: B, B), and $B_2$ be the event of selecting Bag 2 (one black, one white ball: B, W).  
Let $E_1$ represent drawing a black ball first, and $E_2$ represent drawing a black ball second (from the same bag).  
Therefore, we need to find $\mathbb{P}(E_2 | E_1)$.

First, compute the probability of drawing a black ball first:
$$\mathbb{P}(E_1) = \mathbb{P}(E_1|B_1)\times\mathbb{P}(B_1) + \mathbb{P}(E_1|B_2)\times\mathbb{P}(B_2)$$
Since $\mathbb{P}(B_1) = \mathbb{P}(B_2) = \frac{1}{2}$,  
and $\mathbb{P}(E_1|B_1) = 1$ (both balls are black in Bag 1),  
while $\mathbb{P}(E_1|B_2) = \frac{1}{2}$ (only one black ball in Bag 2),  
we have:
$$\mathbb{P}(E_1) = 1\times\frac{1}{2} + \frac{1}{2}\times\frac{1}{2} = \frac{1}{2} + \frac{1}{4} = \frac{3}{4}$$

Next, compute $\mathbb{P}(E_2 | E_1)$:
$$\mathbb{P}(E_2 | E_1) = \frac{\mathbb{P}(E_2 \cap E_1)}{\mathbb{P}(E_1)}$$
The event $E_2 \cap E_1$ means both drawn balls are black.  
Using the law of total probability:
$$\mathbb{P}(E_2 \cap E_1) = \mathbb{P}(E_2 \cap E_1 | B_1)\times\mathbb{P}(B_1) + \mathbb{P}(E_2 \cap E_1 | B_2)\times\mathbb{P}(B_2)$$
If Bag 1 is chosen, both balls are black, so $\mathbb{P}(E_2 \cap E_1 | B_1) = 1$.  
If Bag 2 is chosen, it is impossible to draw two black balls, so $\mathbb{P}(E_2 \cap E_1 | B_2) = 0$.  
Thus:
$$\mathbb{P}(E_2 \cap E_1) = 1\times\frac{1}{2} + 0\times\frac{1}{2} = \frac{1}{2}$$

Finally:
$$\mathbb{P}(E_2 | E_1) = \frac{1/2}{3/4} = \frac{1}{2} \times \frac{4}{3} = \frac{2}{3}$$

=> The probability that the second ball is black given that the first was black is: **[d] 2/3**.


## **Question 4:**

The sample size is $N = 10$:
- The probability of drawing a red marble is $\mu = 0.55$.
- The probability of drawing a green (non-red) marble is $1 - \mu = 1 - 0.55 = 0.45$.

We want the probability of getting no red marbles, $\nu=0$, which means all 10 marbles are green. Since the draws are independent, this follows a Bernoulli trial/Binomial distribution where $k=0$ successes (red marbles) are drawn in $N=10$ trials.$$\mathbb{P}(\nu=0) = \mathbb{P}(\text{10 Green}) = (1-\mu)^{10} = (0.45)^{10}$$


```python
# Constants
prob_red = 0.55
prob_green = 1 - prob_red
num_draws = 10

# Given options
options = {"a": 7.331e-6, "b": 3.405e-4, "c": 0.289, "d": 0.450, "e": 0.550}

# Compute the probability of drawing no red marbles
prob_all_green = prob_green ** num_draws

# Calculate absolute errors and sort them
errors = {key: abs(value - prob_all_green) for key, value in options.items()}
sorted_options = sorted(errors.items(), key=lambda x: x[1])

# Display result
print(f"Probability of getting no red marbles: {prob_all_green}\n")
print("Closest options (sorted by error):")
for key, error in sorted_options:
    print(f"[{key}] = {options[key]}  ->  error = {error}")

# Highlight the best match
best_option = sorted_options[0][0]
print(f"\n=> The best matching option is: [{best_option}] = {options[best_option]}")
```

=> The closest answer is: **[b] $3.405 \times 10^{-4}$**

## **Question 6:**

The score calculation depends on how many of the $2^3=8$ possible target functions ($f$) agree with a given hypothesis ($g$) on the three out-of-sample points: 101, 110, and 111.

Since the 8 possible target functions represent all $2^3$ possible labelings of these three points, for **any fixed hypothesis $g$**, the count of agreements with these target functions follows a fixed combinatoric distribution:

1.  **Agrees on all 3 points:** Only 1 target function perfectly matches $g$. $\binom{3}{3} = 1$
2.  **Agrees on exactly 2 points:** Choose which 2 of the 3 points match $g$. $\binom{3}{2} = 3$
3.  **Agrees on exactly 1 point:** Choose which 1 of the 3 points matches $g$. $\binom{3}{1} = 3$
4.  **Agrees on 0 points:** Only 1 target function is the exact opposite of $g$. $\binom{3}{0} = 1$

The final score for any hypothesis $g$ is therefore:
$$\text{Score} = (1) \times 3 + (3) \times 2 + (3) \times 1 + (1) \times 0 = 3 + 6 + 3 + 0 = \mathbf{12}$$

Since all hypotheses in options [a] through [d] result in the same score of 12, they are all equivalent in terms of this measure.

=> The correct answer is: **[e] They are all equivalent (equal scores for $g$ in [a] through [d])**.

## *Perceptron Learning Algorithm (PLA)*

### Experiment Setup

* **Initialization**:
  Start the Perceptron Learning Algorithm (PLA) using the weight vector obtained from **Linear Regression** as the initial weights.

* **Training Data**:
  Use $N = 10$ randomly generated data points within the input space
  $X = [-1, 1] \times [-1, 1] $, labeled according to a random target function $f$.

* **PLA Process**:
  At each iteration, the algorithm randomly selects one **misclassified point** and updates the weight vector. The process continues until **all points are correctly classified**.

* **Goal**:
  Measure the **average number of iterations** required for PLA to converge over **1000 runs**.



```python
import numpy as np
```

```python
def generate_target_w():
    """
    Generates target_w from two random, uniformly distributed points in [-1, 1] x [-1, 1].

    Returns
    -------
    target_w : numpy array, shape (3, 1)
        The vector of parameters of f.
    """
    # Generate two points from a uniform distribution over [-1, 1]x[-1, 1]
    p1 = np.random.uniform(-1, 1, 2)
    p2 = np.random.uniform(-1, 1, 2)
    # Compute the target W from these two points
    target_w = np.array([p1[1] * p2[0] - p1[0] * p2[1], p2[1] - p1[1], p1[0] - p2[0]]).reshape((-1, 1))

    return target_w
```

```python
def generate_data(N, target_w):
    """
    Generates a data set by generating random inputs and then using target_w to generate the
    corresponding outputs.

    Parameters
    ----------
    N : int
        The number of examples.
    target_w : numpy array, shape (3, 1)
        The vector of parameters of f.

    Returns
    -------
    X : numpy array, shape (N, 3)
        The matrix of input vectors (each row corresponds to an input vector); the first column of
        this matrix is all ones.
    Y : numpy array, shape (N, 1)
        The vector of outputs.
    """
    X = np.random.uniform(-1, 1, (N, 2))
    X = np.hstack((np.ones((N, 1)), X)) # Add 'ones' column
    Y = np.sign(np.dot(X, target_w))

    return X, Y
```

```python
def run_PLA(X, Y):
    """
    Runs PLA.

    Parameters
    ----------
    X : numpy array, shape (N, 3)
        The matrix of input vectors (each row corresponds to an input vector); the first column of
        this matrix is all ones.
    Y : numpy array, shape (N, 1)
        The vector of outputs.

    Returns
    -------
    w : numpy array, shape (3, 1)
        The vector of parameters of g.
    num_iterations : int
        The number of iterations PLA takes to converge.
    """
    w = np.zeros((X.shape[1], 1)) # Init w
    iteration = 0

    # TODO
    while True:
        misclassified = [] # List of misclassified points
        for i in range(X.shape[0]):
            # Check if the point is misclassified
            if np.sign(np.dot(X[i], w)) != Y[i]:
                misclassified.append(i)
                w += Y[i] * X[i].reshape(-1, 1) # Update w
                iteration += 1

        # If no point is misclassified, stop the loop
        if len(misclassified) == 0:
            break

    return w, iteration
```

```python
# Given Options
options_ques7 = {"a": 1, "b": 15, "c": 300, "d": 5000, "e": 10000}
options_ques8 = {"a": 0.001, "b": 0.01, "c": 0.1, "d": 0.5, "e": 0.8}

options_ques9 = {"a": 50, "b": 100, "c": 500, "d": 1000, "e": 5000}
options_ques10 = {"a": 0.001, "b": 0.01, "c": 0.1, "d": 0.5, "e": 0.8}
```

```python
def main(N):
    """
    Parameters
    \t-
    N : int
        The number of training examples.
    """
    num_runs = 1000
    avg_num_iterations = 0.0 # The average number of iterations PLA takes to converge
    avg_test_err = 0.0 # The average test error of g - the final hypothesis picked by PLA

    for r in range(num_runs):
        # Generate target_w
        target_w = generate_target_w()

        # Generate training set
        X, Y = generate_data(N, target_w)

        # Run PLA to pick g
        w, num_iterations = run_PLA(X, Y)

        # Generate test set
        X_test, Y_test = generate_data(10000, target_w)

        # Test g
        test_err = np.mean(np.sign(np.dot(X_test, w)) != Y_test)

        # Update average values
        avg_num_iterations += (num_iterations * 1.0 / num_runs)
        avg_test_err += (test_err * 1.0 / num_runs)
        
    print(f"\t Final Result")
    print(f"avg_num_iterations = {avg_num_iterations:.3f}")
    print(f"avg_test_err = {avg_test_err:.3f}")

    if N == 10:
        options_iter = options_ques7
        options_error = options_ques8
    elif N == 100:
        options_iter = options_ques9
        options_error = options_ques10

    print("\n\t For Iterations")
    errors = {key: abs(value - avg_num_iterations) for key, value in options_iter.items()}
    sorted_options = sorted(errors.items(), key=lambda x: x[1])

    print("Closest options (sorted by error):")
    for key, error in sorted_options:
        print(f"[{key}] = {options_iter[key]:.3f}  ->  error = {error:.3f}")

    best_option = sorted_options[0][0]
    print(f"\n=> The best matching option is: [{best_option}] = {options_iter[best_option]:.3f}")


    print("\n\t For Test Error (E_out)")
    errors = {key: abs(value - avg_test_err) for key, value in options_error.items()}
    sorted_options = sorted(errors.items(), key=lambda x: x[1])

    print("Closest options (sorted by error):")
    for key, error in sorted_options:
        print(f"[{key}] = {options_error[key]:.3f}  ->  error = {error:.3f}")

    best_option = sorted_options[0][0]
    print(f"\n=> The best matching option is: [{best_option}] = {options_error[best_option]:.3f}")
```
